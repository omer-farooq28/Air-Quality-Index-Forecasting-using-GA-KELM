```python
def interpolate_nans(X):
    """Overwrite NaNs with column-wise linear interpolation."""
    X = np.asarray(X, dtype=float).copy()
    for j in range(X.shape[1]):
        mask_j = np.isnan(X[:, j])
        if not mask_j.any():
            continue
        if mask_j.all():
            X[:, j] = 0.0 
            continue
        X[mask_j, j] = np.interp(np.flatnonzero(mask_j),
                                 np.flatnonzero(~mask_j),
                                 X[~mask_j, j])
    return X
```

```python
TARGET = "PM2.5"

LEAKY_OR_NON_NUMERIC = ["City", "Date", "AQI", "AQI_Bucket", TARGET]

feature_cols = [c for c in dataset.columns if c not in LEAKY_OR_NON_NUMERIC]
print("Target   :", TARGET)
print("Features :", feature_cols)


data = dataset.dropna(subset=[TARGET]).reset_index(drop=True)
print(f"\nDropped {len(dataset) - len(data)} rows with a missing target.")

X = interpolate_nans(data[feature_cols].to_numpy(dtype=float))
Y = data[[TARGET]].to_numpy(dtype=float)

print("X shape:", X.shape, "| Y shape:", Y.shape)
assert not np.isnan(X).any(), "NaNs remain in X"
```

```python
X_train_raw, X_test_raw, y_train_raw, y_test_raw = train_test_split(
    X, Y, test_size=0.2, random_state=RANDOM_STATE
)

sc1 = MinMaxScaler(feature_range=(0, 1))  
sc2 = MinMaxScaler(feature_range=(0, 1)) 

X_train = sc1.fit_transform(X_train_raw)
X_test = sc1.transform(X_test_raw)

y_train = sc2.fit_transform(y_train_raw)
y_test = sc2.transform(y_test_raw)

print("Total records found in dataset  = " + str(X.shape[0]))
print("Total features found in dataset = " + str(X.shape[1]))
print("80% dataset for training : " + str(X_train.shape[0]))
print("20% dataset for testing  : " + str(X_test.shape[0]))
```

```python
results_store = {}


def calculateMetrics(algorithm, predict, test_labels, scaler=None):
    """Report MSE/RMSE in normalized space and, if a scaler is given, in real units."""
    predict = np.asarray(predict, dtype=float).reshape(-1, 1)
    test_labels = np.asarray(test_labels, dtype=float).reshape(-1, 1)

    mse_value = mean_squared_error(test_labels, predict)
    rmse_value = float(np.sqrt(mse_value))

    row = {"MSE": mse_value, "RMSE": rmse_value}
    print(f"{algorithm} MSE  : {mse_value:.6f}")
    print(f"{algorithm} RMSE : {rmse_value:.6f}")

    if scaler is not None:
        pred_u = scaler.inverse_transform(predict)
        true_u = scaler.inverse_transform(test_labels)
        rmse_u = float(np.sqrt(mean_squared_error(true_u, pred_u)))
        row["RMSE (target units)"] = rmse_u
        print(f"{algorithm} RMSE : {rmse_u:.4f} (original units)")

    results_store[algorithm] = row
```

```python

svm_cls = SVR()
svm_cls.fit(X_train, y_train.ravel())
predict = svm_cls.predict(X_test)
calculateMetrics("SVR", predict, y_test, scaler=sc2)
```

```python
elm = GeneticELMRegressor()
elm.fit(X_train, y_train.ravel())
predict = elm.predict(X_test)
calculateMetrics("GA-KELM", predict, y_test, scaler=sc2)
```

```python
# Load complete historical dataset
forecast_data = pd.read_csv("../dataset/Dataset.csv")

# Convert Date column
forecast_data["Date"] = pd.to_datetime(forecast_data["Date"], errors="coerce")

# Sort chronologically
forecast_data = forecast_data.sort_values("Date").reset_index(drop=True)

# Remove rows where AQI is unavailable
forecast_data = forecast_data.dropna(subset=["AQI"]).reset_index(drop=True)


# Create time-series features
# Number of previous days used to predict the next day
LAG_DAYS = 14

# Create lag features
for lag in range(1, LAG_DAYS + 1):
    forecast_data[f"AQI_Lag_{lag}"] = forecast_data["AQI"].shift(lag)

# Rolling statistics
forecast_data["AQI_Rolling_Mean_7"] = (
    forecast_data["AQI"].shift(1).rolling(7).mean()
)

forecast_data["AQI_Rolling_Mean_14"] = (
    forecast_data["AQI"].shift(1).rolling(14).mean()
)

# Date-based features
forecast_data["DayOfYear"] = forecast_data["Date"].dt.dayofyear
forecast_data["Month"] = forecast_data["Date"].dt.month

# Cyclic date features
forecast_data["Sin_Day"] = np.sin(
    2 * np.pi * forecast_data["DayOfYear"] / 365.25
)

forecast_data["Cos_Day"] = np.cos(
    2 * np.pi * forecast_data["DayOfYear"] / 365.25
)


# Prepare training data
feature_cols = (
    [f"AQI_Lag_{i}" for i in range(1, LAG_DAYS + 1)]
    + [
        "AQI_Rolling_Mean_7",
        "AQI_Rolling_Mean_14",
        "Sin_Day",
        "Cos_Day"
    ]
)

model_data = forecast_data.dropna(
    subset=feature_cols + ["AQI"]
).reset_index(drop=True)

X_forecast = model_data[feature_cols].values
y_forecast = model_data[["AQI"]].values

# Scale features and target
forecast_X_scaler = MinMaxScaler()
forecast_y_scaler = MinMaxScaler()

X_scaled = forecast_X_scaler.fit_transform(X_forecast)
y_scaled = forecast_y_scaler.fit_transform(y_forecast)

# Train GA-KELM

forecast_elm = GeneticELMRegressor()

forecast_elm.fit(
    X_scaled,
    y_scaled.ravel()
)

print("GA-KELM forecasting model trained successfully.")
```

```python

# Prediction
FORECAST_DAYS = 10

aqi_history = forecast_data["AQI"].tolist()

last_date = forecast_data["Date"].iloc[-1]

future_dates = pd.date_range(
    start=last_date + pd.Timedelta(days=1),
    periods=FORECAST_DAYS,
    freq="D"
)

future_predictions = []

for future_date in future_dates:

    lag_values = aqi_history[-LAG_DAYS:]
    lag_values = lag_values[::-1]

    rolling_7 = np.mean(aqi_history[-7:])
    rolling_14 = np.mean(aqi_history[-14:])

    day_of_year = future_date.dayofyear

    sin_day = np.sin(
        2 * np.pi * day_of_year / 365.25
    )

    cos_day = np.cos(
        2 * np.pi * day_of_year / 365.25
    )

    feature_row = (
        lag_values
        + [
            rolling_7,
            rolling_14,
            sin_day,
            cos_day
        ]
    )

    feature_row = np.array(feature_row).reshape(1, -1)

    feature_scaled = forecast_X_scaler.transform(feature_row)

    prediction_scaled = forecast_elm.predict(feature_scaled)

    prediction = forecast_y_scaler.inverse_transform(
        np.array(prediction_scaled).reshape(-1, 1)
    )[0, 0]

    prediction = max(0, prediction)

    future_predictions.append(prediction)

    aqi_history.append(prediction)


forecast_result = pd.DataFrame({
    "Date": future_dates,
    "Forecasted_AQI": future_predictions
})

for i, row in forecast_result.iterrows():
    print(
        f"Date = {row['Date'].date()} "
        f"=====> Forecasted AQI : {row['Forecasted_AQI']:.2f}"
    )
```
